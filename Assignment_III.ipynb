The project problem is related to the high power consumption of cooling tower fans for the geothermal power plants, especially during off-peak periods and when the ambient temperatures are low because the cooling process is highly dependent on local ambient conditions.
Therefore the intended AI/ML outcome is an objective function that optimizes power consumption of the induced draft cooling tower fans for the whole range of plant operation loads and takes into consideration the variable ambient conditions.

Input Features
Environmental Inputs: Ambient air temperature, relative humidity, wet-bulb temperature.
Operational Inputs: Hot water temperature, mass/volume flow rate, fan speed/frequency and power consumption of fans and pump.
Target Outputs: Cold water temperature (Outlet), range, approach, evaporation losses.
