```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    """Converts a temperature from Celsius to Fahrenheit."""
    return (celsius * 9 / 5) + 32


# Main program loop
def main():
    # Define a threshold temperature in Fahrenheit ( room temp cutoff)
    THRESHOLD_F = 77.0  # 77°F is approximately 25°C

    print("--- Temperature Converter ---")
    print(f"Threshold set to: {THRESHOLD_F}°F\n")

    while True:
        # Prompt user for input
        user_input = input(
            "Enter temperature in °C (or type 'exit' to stop): "
        )

        # Check if the user wants to quit the loop
        if user_input.lower() == "exit":
            print("Program ended. Goodbye!")
            break

        try:
            # Convert string input to a floating-point number
            celsius = float(user_input)

            # Call the conversion function
            fahrenheit = celsius_to_fahrenheit(celsius)

            # Display converted temperature
            print(f"{celsius}°C = {fahrenheit:.2f}°F")

            # Compare converted temperature to the threshold
            if fahrenheit > THRESHOLD_F:
                print(f"-> {fahrenheit:.2f}°F is ABOVE the threshold ({THRESHOLD_F}°F).\n")
            elif fahrenheit < THRESHOLD_F:
                print(f"-> {fahrenheit:.2f}°F is BELOW the threshold ({THRESHOLD_F}°F).\n")
            else:
                print(f"-> {fahrenheit:.2f}°F is EQUAL to the threshold ({THRESHOLD_F}°F).\n")

        except ValueError:
            print("Invalid input. Please enter a numerical value.\n")


# Run the program
if __name__ == "__main__":
    main()
```
